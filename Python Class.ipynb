```python
#Set Operations - Set support powerful operations like union, intersection, difference and symmetric difference

#Union() - combines two sets

set1 = {1, 2, 3}
set2 ={3, 4, 5}
result = set2.union(set1)
print(result)
```

```python
#Alternatively, use the | operator |(pipe symbol)

result = set1 | set2
print(result)
```

```python
#8. Intersection() - common elements between sets
# This methods returns elements present in both sides

set1 = {1, 2, 3, 4}
set2 ={3, 4, 5, 6}
result = set2.intersection(set1)
print(result)
```

```python
#Alternatively, use the & symbol

result = set1 & set2
print(result)
```

```python
#9. Difference() - Elements in one set but not the other 
#This method returs elements that exist in the firs set but not in the second 

set1 = {1, 2, 3, 4}
set2 ={3, 4, 5, 6}
result = set2.difference(set1)
print(result)
```

```python
#Alternatively, use the - operator 

result = set2 - set1
print(result)
```

```python
#10. Symmetric_difference() - Elements in either set but not both
#This method returs elements that are in one of the sets but not in both. It is the opposite of intersection

set1 = {1, 2, 3, 4}
set2 ={3, 4, 5, 6}
result = set2.symmetric_difference(set1)
print(result)

```

```python
#Alternatively, use the ^ operator 

result = set2 ^ set1
print(result)
```

```python
#Set Comparisons

#11. issubset() - Checks if one set is a subset of another. It returs True if all elements of the first set exists in the second set

set1 = {1, 2,}
set2 ={1, 2, 3, 4}
print(set1.issubset(set2))
```

```python
#Alternatively, use <= operator

print(set1<=set2)
```

```python
#12. issuperset() -  checks if one set is a superset of another. It returns True  if the first set contains all elements of the second set

set1 = {1, 2, 3, 4}
set2 ={1, 2}
print(set1.issuperset(set2))
```

```python
#Alternatively, use >= operator

print(set1>=set2)
```
